```python
import pandas as pd

def ingest_gcp_billing(file_path):
    # Ingest the billing export
    df = pd.read_csv(file_path)

    # 1. Data Cleaning: Convert usage/export times to datetime objects
    df['usage_start_time'] = pd.to_datetime(df['usage_start_time'])
    df['usage_end_time'] = pd.to_datetime(df['usage_end_time'])

    # 2. Filtering: Focus on 'Google Cloud Platform' services
    # In a real export, this would filter by Project ID or Service Description
    gcp_expenses = df[df['service_description'].str.contains('Cloud', na=False)]

    # 3. Handling Nulls: Ensure no missing cost values
    gcp_expenses = gcp_expenses.dropna(subset=['cost'])

    return gcp_expenses
```

def map_to_ledger(billing_df):
    # Mapping actual costs to the '2025 Annual Budget' metadata
    # Source [1] identifies GCP is paid via Charles Schwab on the 1st
    billing_df['payment_source'] = 'Charles Schwab'
    billing_df['budgeted_amount'] = 1.00

    # Calculate Variance (Actual vs. Budgeted)
    summary = billing_df.groupby('service_description')['cost'].sum().reset_index()
    summary['variance'] = summary['cost'] - 1.00
    
    return summary
